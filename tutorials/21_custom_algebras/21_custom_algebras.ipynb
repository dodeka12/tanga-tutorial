# 21. Custom Algebras and Advanced Patterns

Go beyond the four built-in basis classes. Construct custom algebras with arbitrary dimension and signature (up to 31 dimensions). Show how to assign names to blades manually, create ad-hoc basis classes, and integrate custom algebras with BladeMask, the solver pipeline, and the geometry submodule (where supported). Cover sparse blade encoding, performance considerations, and the constexpr-friendly C++ template engine behind the scenes.
