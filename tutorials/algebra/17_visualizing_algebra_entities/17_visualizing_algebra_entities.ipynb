# Visualizing Algebra Entities — Full Coverage

**Part II · Geometric Algebra & Core** — Tutorial 17

This chapter maps the geometry data model from [Tutorial 15](../15_geometry/)
into the 3D viewer. The viewer accepts an entity/operator or a raw multivector;
raw multivectors are analyzed on the way in, honoring the algebra's `opns` flag.

By the end you will be able to:

- Feed `geo(entity)` results (and plain `MV`s) into `viz.add(...)`.
- Render points, directions, planes, and space in E3.
- Render spheres, circles, point pairs, and lines in N3.
- Render rotors, translators, dilators, inversions, and reflections.
- Render plane-based lines, planes, translators, and point reflections in PGA3.
- Export self-contained HTML snapshots with `export_snapshot`.

> **Prerequisites:** [Tutorial 15](../15_geometry/) (the geometry pipeline). The
> viewer API itself (scenes, styles, animation, export) is covered in
> [Part I — Visualization](../../visualization/).

## 1. Setup

Bind three cached basis algebras — E3 (Euclidean), N3 (conformal), and PGA3
(plane-based) — each to a `Geometry`, and import the `Visualizer`.

```python
import math
import os

from pytanga.basis import BasisE3, BasisN3, BasisPGA3
from pytanga.geometry import (
    Circle, Dilator, Direction, Geometry, Inversion, Line, Plane, Point,
    PointPair, ReflectionPlane, ReflectionPoint, Rotor, Space, Sphere, Translator,
)
from pytanga.viz import Visualizer

E3 = BasisE3()
N3 = BasisN3()
PGA3 = BasisPGA3()

geo_e3 = Geometry(E3)
geo_n3 = Geometry(N3)
geo_pga3 = Geometry(PGA3)
```

## 2. Mapping entities to multivectors

`geo(entity)` creates the MV for each entity/operator. The viewer accepts either
the entity (creating it itself) or the raw MV (analyzing it on the way in).

```python
print("E3 point MV :", geo_e3(Point(1, 1, 1)).to_dict())
print("N3 sphere MV:", geo_n3(Sphere(center=Point(0, 0, 0), radius=2.0)).to_dict())
print("PGA3 line MV:", geo_pga3(Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0))).to_dict())
```

## 3. E3 entities in the viewer

A grade-1 vector renders as a direction, a grade-2 bivector as a plane, and the
pseudoscalar as space.

```python
viz_e3 = Visualizer(title="E3 — point, direction, plane, space")
viz_e3.add(geo_e3(Point(1, 1, 1)), color="#ffcc00", label="point")
viz_e3.add(geo_e3(Direction(1, 0, 0)), color="#44ff44", label="direction")
viz_e3.add(geo_e3(Plane(point=Point(0, 0, 0), normal=Direction(0, 0, 1))), color="#4488ff", opacity=0.3, label="plane")
viz_e3.add(geo_e3(Space()), color="#888888", label="space")

os.makedirs("_output/17_visualizing_algebra_entities", exist_ok=True)
viz_e3.export_snapshot("_output/17_visualizing_algebra_entities/entities_e3.html", overwrite=True)
print("exported entities_e3.html")
```

## 4. E3 operators in the viewer

Operators render the same way: a `Rotor` shows a rotation (a displaced arc) and
a `ReflectionPlane` shows its plane.

```python
viz_e3ops = Visualizer(title="E3 — rotor and reflection plane")
viz_e3ops.add(geo_e3(Rotor(math.pi / 4, Direction(0, 0, 1))), color="#ffcc00", label="rotor")
viz_e3ops.add(geo_e3(ReflectionPlane(Direction(1, 0, 0))), color="#44ff44", opacity=0.3, label="reflection plane")

viz_e3ops.export_snapshot("_output/17_visualizing_algebra_entities/operators_e3.html", overwrite=True)
print("exported operators_e3.html")
```

## 5. N3 conformal entities in the viewer

Conformal N3 adds the round entities: a **sphere** (grade-4), a **circle**
(sphere ∩ plane, grade-3), and a **point pair** (grade-2). A **line** is the
grade-3 blade of two points and infinity.

```python
viz_n3 = Visualizer(title="N3 — sphere, circle, point pair, line")
viz_n3.add(geo_n3(Sphere(center=Point(0, 0, 0), radius=2.0)), color="#4488ff", opacity=0.35, label="sphere")
viz_n3.add(geo_n3(Circle(center=Point(3, 0, 0), radius=1.5, normal=Direction(0, 0, 1))), color="#ff44ff", label="circle")
viz_n3.add(geo_n3(PointPair(point_a=Point(-2, 0, 0), point_b=Point(-1, 0, 0))), color="#44ff44", label="point pair")
viz_n3.add(geo_n3(Line(origin=Point(0, 0, 0), direction=Direction(0, 1, 0))), color="#ffcc00", label="line")

viz_n3.export_snapshot("_output/17_visualizing_algebra_entities/entities_n3.html", overwrite=True)
print("exported entities_n3.html")
```

## 6. N3 conformal operators in the viewer

N3 adds the conformal operators: `Translator`, `Dilator`, and `Inversion`
(inversion in a sphere), alongside the `Rotor`.

```python
viz_n3ops = Visualizer(title="N3 — rotor, translator, dilator, inversion")
viz_n3ops.add(geo_n3(Rotor(math.pi / 4, Direction(0, 0, 1))), color="#ffcc00", label="rotor")
viz_n3ops.add(geo_n3(Translator(vector=Direction(1, 0, 0))), color="#44ff44", label="translator")
viz_n3ops.add(geo_n3(Dilator(factor=2.0)), color="#ff44ff", label="dilator")
viz_n3ops.add(geo_n3(Inversion(center=Point(0, 0, 0))), color="#4488ff", label="inversion")

viz_n3ops.export_snapshot("_output/17_visualizing_algebra_entities/operators_n3.html", overwrite=True)
print("exported operators_n3.html")
```

## 7. PGA3 plane-based entities and operators

PGA3 maps planes to grade-1 vectors and points to grade-3 trivectors. Here a
line (grade-2), a plane, a translator, and a point reflection are rendered.

```python
viz_pga3 = Visualizer(title="PGA3 — line, plane, translator, point reflection")
viz_pga3.add(geo_pga3(Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0))), color="#ffcc00", label="line")
viz_pga3.add(geo_pga3(Plane(point=Point(0, 0, 1), normal=Direction(0, 0, 1))), color="#4488ff", opacity=0.3, label="plane")
viz_pga3.add(geo_pga3(Translator(vector=Direction(0, 1, 0))), color="#44ff44", label="translator")
viz_pga3.add(geo_pga3(ReflectionPoint(Point(0, 0, 0))), color="#ff44ff", label="point reflection")

viz_pga3.export_snapshot("_output/17_visualizing_algebra_entities/pga3.html", overwrite=True)
print("exported pga3.html")
```

## 8. Raw multivectors are analyzed on the way in

A raw multivector is analyzed before rendering, honoring `algebra.opns`. Here a
grade-2 bivector is read as a plane (OPNS).

```python
raw = E3("2 e12")          # a grade-2 bivector

viz_raw = Visualizer(title="E3 — a raw bivector analyzed into a plane")
viz_raw.add(raw, color="#4488ff", opacity=0.3, label="raw e12 bivector")

viz_raw.export_snapshot("_output/17_visualizing_algebra_entities/raw_mv.html", overwrite=True)
print("exported raw_mv.html")
```

## 9. Notes and next steps

- **Coverage.** The examples above cover E3 (points, directions, planes, space,
  rotors, reflections), N3 (spheres, circles, point pairs, lines, rotors,
  translators, dilators, inversions), and PGA3 (lines, planes, translators,
  point reflections). The same `geo(...)` / `viz.add(...)` pattern applies
  unchanged to P3 and the 2D models.
- **`Motor` and `GeneralRotor`.** These round-trip through the geometry
  pipeline (see [Tutorial 15](../15_geometry/)), but the viewer cannot render
  them directly in this release — they analyze to a `GeneralRotor`, which the
  renderer does not yet draw. Render their `Rotor` / `Translator` parts
  separately instead.
- **Viewer tooling** (scenes, axes/grid/camera, labels, interaction, animation,
  export) is covered in [Part I — Visualization](../../visualization/).

| Task | API |
|---|---|
| Entity → MV | `geo(entity)` |
| MV → viewer (auto-analyzed) | `viz.add(mv, ...)` |
| Entity → viewer | `viz.add(entity, ...)` |
| Export standalone HTML | `viz.export_snapshot(path, overwrite=True)` |
