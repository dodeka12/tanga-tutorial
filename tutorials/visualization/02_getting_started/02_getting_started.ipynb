# Getting Started with the Viewer

**Part I · Visualization** — Tutorial 02

Create a `Visualizer`, add a few plain entities, and open the interactive
Three.js viewer with `show()` — the single display entry point. By the end you
will be able to:

- Create a 3D viewer (default) or a 2D viewer (`space_dim=2`).
- Add `Point`, `Line`, and `Sphere` entities with `add()`.
- Open the viewer with `show()`, and choose between the blocking and
  non-blocking run modes.
- Render inline in a notebook with `display()`.

> **Prerequisites:** none — geometry dataclasses are used as plain 3D data.


## Setup


```python
from pytanga.geometry import Direction, Line, Point, Sphere
from pytanga.viz import SphereStyle, Visualizer
```

## 1. Creating a Visualizer

`Visualizer()` creates a **3D** viewer by default. Pass `space_dim=2` for a 2D
viewer (an orthographic top-down camera). The title changes accordingly
(`"Tanga 3D Viewer"` vs `"Tanga 2D Viewer"`) unless you override it.


```python
viz3d = Visualizer(title="Tanga — 3D Viewer")   # 3D (default)
viz2d = Visualizer(space_dim=2, title="Tanga — 2D Viewer")

print("created two viewers")
```

## 2. Adding entities

`add()` places an entity in the main scene and returns its entity id string
(use the id later to update or remove it). Color/opacity shortcuts and per-type
styles (`SphereStyle`, `PointStyle`, …) are accepted directly.


```python
viz = Visualizer(title="Tanga — Entities")

p_id = viz.add(Point(1, 2, 3), color="#ff4444", label="$P$")
l_id = viz.add(
    Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0)),
    color="#44ff44",
    label="x-axis",
)
s_id = viz.add(
    Sphere(Point(0, 0, 0), 2.0),
    style=SphereStyle(wireframe=True),
    opacity=0.3,
    label="$S$",
)

print("entity ids:", p_id, l_id, s_id)
viz.display_snapshot()
```

## 3. The display entry point — `show()`

`show()` is the single entry point that serves the viewer and shows it:

- In **Jupyter**, `show()` renders inline (delegating to `display()`).
- In a **script**, `show()` opens a browser tab (then `wait()` keeps the script
  alive).

`show()` accepts a `jupyter` option to force either behaviour (`None`
auto-detects, `True` forces inline, `False` forces a browser tab).


```python
viz = Visualizer(title="Tanga — Getting Started")
viz.add(Point(1, 2, 3), color="#ff4444", label="$P$")
viz.add(Sphere(Point(0, 0, 0), 2.0), style=SphereStyle(wireframe=True), opacity=0.3)

viz.show()  # inline in Jupyter, browser tab in a script
# viz.show(jupyter=True)   # force inline
# viz.show(jupyter=False)  # force a browser tab
```

## 4. Run modes

**Blocking** (`show()` + `wait()`) is for plain scripts: `show()` opens the
viewer, `wait()` blocks until Ctrl+C, then stops the server.

**Non-blocking** (`start_server()` / `flush()` / `stop_server()`) is for
notebooks and animation loops: `start_server()` serves without opening a
browser, `flush()` pushes scene changes, and `stop_server()` releases the port.


```python
# Non-blocking mode (notebooks, animation loops):
viz = Visualizer(title="Tanga — Non-blocking")
viz.start_server()                 # serve only (default port 8765)
pid = viz.add(Point(1, 2, 3), color="#ff4444")
viz.flush()                        # push the scene to connected viewers
print("serving at", viz.url)

viz.stop_server()                  # release the port / background thread
print("stopped")

# Blocking mode (plain scripts):
#   viz.show()    # open the viewer
#   viz.wait()    # block until Ctrl+C, then stop the server
```

## 5. Inline Jupyter display — `display()`

`display()` renders the viewer inline (or returns an HTML string outside
Jupyter). Like `show()`, it is **idempotent within a cell** — repeated calls
flush into the already-open viewer rather than opening a second one. The viewer
is keyed by `viewer_name`, the notebook cell id, or the scene name. If called
before the server is running, it prints a hint to call `start_server()`.


```python
viz = Visualizer(title="Tanga — Inline display")
viz.add(Point(1, 2, 3), color="#ff4444", label="$P$")

viz.display()                      # idempotent inline display
viz.add(Point(4, 5, 6), color="#44ff44")
viz.display()                      # flushes into the already-open viewer
```

## 6. Browser tabs & shortcuts

- `reuse_existing=True` (default) waits for an already-open tab to reconnect
  before opening a new one; `reuse_existing=False` opens a fresh tab per
  `show()`.
- `show(wait_for_browser=True)` blocks until a browser actually connects.
- In the live viewer, `Ctrl+S` / `Cmd+S` downloads a PNG snapshot of the
  viewport (the WebGL canvas), and `r` toggles camera auto-rotation.

`show()` / `run()` also accept a `layout=` to open a split-view page — see
[Tutorial 12](../12_split_views/).


## Visual Examples

A standalone HTML figure for each mode (3D and 2D), exported via
`export_snapshot()`.


```python
# 3D figure
viz = Visualizer(title="Tanga — 3D Figure")
viz.add(Point(2, 1, 0), color="#ff4444", label="$P$")
viz.add(
    Sphere(Point(0, 0, 0), 2.0),
    style=SphereStyle(wireframe=True),
    opacity=0.3,
    label="$S$",
)
viz.export_snapshot("_output/02_3d_figure.html", overwrite=True)
print("3D figure written")

# 2D figure (orthographic top-down camera)
viz2 = Visualizer(space_dim=2, title="Tanga — 2D Figure")
viz2.add(Point(3, 4, 0), color="#ff4444", label="$P(3,4)$")
viz2.add(Point(-2, 1, 0), color="#44ff44", label="$Q(-2,1)$")
viz2.export_snapshot("_output/02_2d_figure.html", overwrite=True)
print("2D figure written")
```

## Summary

| Task | API |
|---|---|
| 3D viewer | `Visualizer()` |
| 2D viewer | `Visualizer(space_dim=2)` |
| Add an entity | `viz.add(entity, color=..., opacity=..., style=...)` |
| Open the viewer | `viz.show()` |
| Inline notebook display | `viz.display()` / `viz.display_snapshot()` |
| Blocking run | `viz.show()` then `viz.wait()` |
| Non-blocking run | `viz.start_server()` / `viz.flush()` / `viz.stop_server()` |
