# Visualizing GA Entities and Operators

**Part I · Visualization** — Tutorial 20

A basic bridge from visualization to geometric algebra. You will learn to:

- Feed raw multivectors into `viz.add(...)` — they are analyzed into entities
  on the way in, honoring the algebra's `opns` flag.
- Render operators (`Rotor`, `Translator`, `Motor`, `Reflection`, `Inversion`,
  `Dilator`).
- Understand what `GeneralRotor` and `Motor` draw.

> **High-level only** — this chapter describes *what* gets drawn and how to
> point it at the viewer. For the meaning of the underlying blades, OPNS/IPNS,
> and how to construct entities/operators, see
> [Part II — Geometric Algebra](../../algebra/).


## Setup


```python
import math

from pytanga.basis import BasisE3, BasisN3
from pytanga.geometry import (
    Circle, Direction, Dilator, GeneralRotor, Geometry, Inversion, Line, Motor,
    Point, ReflectionPlane, Rotor, Sphere, Translator,
)
from pytanga.viz import Visualizer
```

## 1. Raw multivectors → the viewer

`viz.add(mv)` accepts a raw multivector and analyzes it into an entity on the
way in, honoring the algebra's `opns` flag. Equivalently, pass the entity
directly.


```python
E3 = BasisE3()
geo = Geometry(E3)

# An entity → MV, then the MV → the viewer (auto-analyzed):
mv_point = geo(Point(1, 1, 1))
print("point MV:", mv_point.to_dict())

viz = Visualizer(title="GA — multivector input", add_default_axes=False, add_default_grid=False)
viz.add(mv_point, color="#ff4444", label="$P$")
viz.add(geo(Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0))), color="#44ff44", label="$L$")
viz.display_snapshot()
```

## 2. Rendering operators

Operators are created through `pytanga.geometry` and fed to the viewer the same
way. The viewer draws a dedicated glyph per operator kind:

- `Rotor` — a disc arc, a torus, and an axis line.
- `Translator` — an arrow along the translation vector.
- `Dilator` — concentric rings.
- `ReflectionPlane` — a translucent plane.
- `Inversion` — a sphere glyph (in conformal algebras).


```python
N3 = BasisN3()
geo_n3 = Geometry(N3)

viz = Visualizer(title="GA — operators", add_default_axes=False, add_default_grid=False)

# (Translator, Motor, GeneralRotor, and Inversion require the conformal N3 algebra.)
viz.add(geo_n3(Rotor(angle=math.pi / 3, axis=Direction(0, 0, 1))), color="#ff8844", label="Rotor")
viz.add(geo_n3(Translator(vector=Direction(2, 0, 0))), color="#44aaff", label="Translator")
viz.add(geo_n3(Dilator(factor=1.5, origin=Point(0, 0, 0))), color="#ffcc44", label="Dilator")
viz.add(geo_n3(ReflectionPlane(plane_or_normal=Direction(1, 0, 0))), color="#88ccff", label="Reflection")
viz.add(geo_n3(Inversion(center=Point(0, 0, 0), radius=1.0)), color="#cc88ff", label="Inversion")

viz.display_snapshot()
```

## 3. `GeneralRotor` and `Motor`

- `GeneralRotor` renders with the **same** rotor visualization as `Rotor`
  (disc arc, torus, and axis line), displaced to its `origin`.
- `Motor` renders a displaced rotation (general rotor) **plus** a translation
  arrow along the screw axis.


```python
N3 = BasisN3()
geo_n3 = Geometry(N3)

viz = Visualizer(title="GA — GeneralRotor & Motor", add_default_axes=False, add_default_grid=False)

gr = GeneralRotor(angle=math.pi / 3, axis=Direction(0, 0, 1), origin=Point(1, 0, 0))
motor = Motor(
    rotor=Rotor(angle=math.pi / 3, axis=Direction(0, 0, 1)),
    translator=Translator(vector=Direction(0, 0, 1)),
)

viz.add(geo_n3(gr), color="#ff9966", label="GeneralRotor")
viz.add(geo_n3(motor), color="#ff66cc", label="Motor")

viz.display_snapshot()
```

## 4. Conformal entities (N3)

In the conformal algebra N3, spheres, circles, and point pairs have direct
multivector representations (OPNS vs IPNS read from `algebra.opns`).


```python
N3 = BasisN3()
geo_n3 = Geometry(N3)

viz = Visualizer(title="GA — conformal entities", add_default_axes=False, add_default_grid=False)

viz.add(geo_n3(Sphere(center=Point(0, 0, 0), radius=2.0)), color="#ffaa00", opacity=0.3, label="$S$")
viz.add(geo_n3(Circle(center=Point(0, 0, 0), normal=Direction(0, 0, 1), radius=2.5)), color="#ff44ff", label="$C$")

viz.display_snapshot()
```

## Visual Examples

A scene mixing entities and operators, exported via `export_snapshot()`.


```python
N3 = BasisN3()
geo = Geometry(N3)

viz = Visualizer(title="GA — entities & operators")
viz.add(geo(Point(0, 0, 0)), color="#ffff00", label="$O$")
viz.add(geo(Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0))), color="#44ff44", label="$L$")
viz.add(geo(Rotor(angle=math.pi / 3, axis=Direction(0, 0, 1))), color="#ff8844", label="Rotor")
viz.add(geo(Translator(vector=Direction(0, 2, 0))), color="#44aaff", label="Translator")
viz.add(
    geo(Motor(rotor=Rotor(angle=math.pi / 3, axis=Direction(0, 0, 1)),
              translator=Translator(vector=Direction(0, 0, 1)))),
    color="#ff66cc", label="Motor",
)

viz.display_snapshot()
```

## Summary

| Task | API |
|---|---|
| Entity → MV | `geo(entity)` |
| MV → viewer (auto-analyzed) | `viz.add(mv, ...)` |
| Entity → viewer | `viz.add(entity, ...)` |
| OPNS/IPNS | read from `algebra.opns` (mutable, default `True`) |
| `Rotor` | disc arc + torus + axis line |
| `GeneralRotor` | `Rotor` visualization displaced to its `origin` |
| `Motor` | displaced rotation + translation arrow along the screw axis |
