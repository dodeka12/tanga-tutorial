# 16. 3D Visualization — Interactive Scenes

Bring geometric algebra to life with pytanga.viz. Create a Visualizer, add entities (Point, Line, Plane, Sphere, Circle, point pairs), apply per-entity styles (color, opacity, wireframe, point size), and launch the interactive Three.js viewer with viz.run(). Cover camera configuration (auto-fit, explicit, partial), orbit controls, and the Ctrl+S screenshot feature. Demonstrate multi-scene support — creating named scenes, side-by-side display in Jupyter with display_row(). Show how to input raw MVs (PGA3, N3) for OPNS vs IPNS visualization.
