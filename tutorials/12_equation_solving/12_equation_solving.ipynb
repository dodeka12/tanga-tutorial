# 12. Equation Solving — From GA to Linear Systems

Teach the solve, solve_lsq, and solve_mod free functions. Start with finding the multiplicative inverse of a multivector (A * X = 1). Move to solving for an unknown multivector in product equations (A ∘ X = Y), over-determined systems with solve_lsq, and modular systems with solve_mod. Explain the automatic blade mask derivation pipeline — how the solver infers the unknown's subspace from the inputs. Cover the EProduct and EInv enums that control which GA product is used and whether involutions are applied.
