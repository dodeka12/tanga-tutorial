# 5. Euclidean 3D (G(3,0)) — Vectors, Bivectors, Rotors

Deep dive into BasisE3. Build vectors and bivectors from coordinates and from the geometric product of basis vectors. Compute the outer product to form bivectors, use the inner product for metric relationships. Construct rotors from angle-axis pairs using alg.rotor(), apply them with the versor/sandwich product R * v * ~R, compose multiple rotations, and verify results. Cover the pseudoscalar, Hodge dual via the signed dual operator, and the correspondence between bivectors and rotation planes.
