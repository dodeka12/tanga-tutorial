# Getting Started with TanGA — 3D Euclidean Geometric Algebra

This tutorial walks through the basics of using TanGA for **3D Euclidean geometric algebra (G(3,0))**.
You'll learn how to:

1. Create an algebra
2. Work with basis vectors and multivectors
3. Perform geometric, inner, and outer products
4. Build and apply rotors for 3D rotations
5. Use duality and grade extraction

## Prerequisites

Make sure you have followed the [setup instructions](../README.md) before running this tutorial.

### 1. Creating an Algebra

TanGA provides several pre-built basis classes. For 3D Euclidean space, use `E3`:

```python
from pytanga import Algebra

# Create a 3D Euclidean algebra with named basis vectors
alg = Algebra.from_name("E3")

# Access basis vectors
e1, e2, e3 = alg.e1, alg.e2, alg.e3
print(f"Algebra: dim={alg.dim}, signature={alg.sig}")
print(f"Basis vectors: {e1}, {e2}, {e3}")
print(f"Algebra dimension (number of blades): {alg.algebra_dim}")
```

The algebra `E3` has $2^3 = 8$ basis blades:
  - **1** scalar (grade 0)
  - **e1, e2, e3** vectors (grade 1)
  - **e12, e31, e23** bivectors (grade 2)
  - **I = e123** pseudoscalar (grade 3)

### 2. Geometric, Inner, and Outer Products

The three fundamental products in geometric algebra:

```python
# Geometric product (*):  e1 * e2 = e1 | e2 + e1 ^ e2
print(f"e1 * e2 = {e1 * e2}")   # pure bivector: e12
print(f"e1 * e1 = {e1 * e1}")   # scalar: 1

# Inner product (|):  contraction
print(f"e1 | e1 = {e1 | e1}")   # scalar: 1
print(f"e1 | e2 = {e1 | e2}")   # zero: orthogonal vectors

# Outer product (^):  wedge
print(f"e1 ^ e2 = {e1 ^ e2}")   # bivector: e12
print(f"e1 ^ e1 = {e1 ^ e1}")   # zero: wedge of a vector with itself
```

### 3. Creating Multivectors

You can create multivectors from dictionaries, string expressions, or coordinates:

```python
# From a dictionary (blade bitmask → coefficient)
# bitmask: 0=scalar, 1=e1, 2=e2, 4=e3, 3=e12, ...
mv = alg.multivector({0: 3, 1: 2, 2: -1})
print(f"From dict: {mv}")

# From a string expression
mv2 = alg.multivector("2.5 + 3 e1 - e23")
print(f"From string: {mv2}")

# Create a vector from coordinates
v = alg.vector(1.0, 2.0, 0.5)
print(f"Vector from coords: {v}")

# Scalar coefficient access
print(f"Scalar part of mv: {mv.scalar}")
```

### 4. Duality — The Cross Product Connection

In 3D Euclidean space, the dual of a bivector is a vector.
This means ★(a ∧ b) = a × b (the familiar cross product):

```python
# Cross product via dual
a = alg.vector(1, 0, 0)
b = alg.vector(0, 1, 0)
cross = (a ^ b).dual()
print(f"★(e1 ^ e2) = {cross}")   # should yield e3 (z-axis)

# Verify: dual(dual(A)) = A (up to sign)
print(f"★★(e1 ^ e2) = {cross.dual()}")
```

### 5. Rotations with Rotors

Rotors are the GA way to represent rotations. A rotor for angle θ around
an axis n̂ is:
R = cos(θ/2) + sin(θ/2) B
where B is the unit bivector orthogonal to the rotation axis.

To rotate a vector v:  v' = R v R̃

```python
from math import pi

# Create a rotor: 90° rotation around e3 axis
rotor = alg.rotor(pi / 2, e3)
print(f"Rotor: {rotor}")
print(f"Rotor magnitude: {rotor.mag:.4f}")

# Rotate e1 by 90° around e3: should become e2
v = e1
v_rotated = rotor * v * rotor.rev()  # versor product
print(f"Rotate e1 by 90° around e3: {v_rotated}")

# Rotate e2 by 90° around e3: should become -e1
v_rotated2 = rotor * e2 * rotor.rev()
print(f"Rotate e2 by 90° around e3: {v_rotated2}")
```

### 6. Grade Extraction and Decomposition

Extract specific grade parts from a multivector:

```python
# Create a mixed-grade multivector
mv = 5.0 + e1 * e2 + 2.0 * e3   # scalar + bivector + vector
print(f"Full mv: {mv}")

# Extract grade parts
print(f"Grade 0 (scalar):  {mv.grade(0)}")
print(f"Grade 1 (vector):  {mv.grade(1)}")
print(f"Grade 2 (bivector): {mv.grade(2)}")
print(f"Grade 3 (trivector): {mv.grade(3)}")

# Check what grades are present
print(f"Grades present: {mv.grades}")

# Check if pure grade
print(f"Is scalar? {mv.is_scalar}")
print(f"Is grade-1? {mv.is_grade(1)}")
print(f"Is zero? {mv.is_zero}")
```

### 7. Magnitude, Normalization, and Inverse

```python
# Magnitude
v = alg.vector(3, 4, 0)
print(f"Vector: {v}")
print(f"Magnitude: {v.mag:.4f}")
print(f"Squared magnitude: {v.mag2}")

# Normalization
v_norm = v.normalized()
print(f"Normalized: {v_norm}")
print(f"Norm magnitude: {v_norm.mag:.4f}")

# Inverse (for a vector: v⁻¹ = v / |v|²)
v_inv = ~v
print(f"Inverse: {v_inv}")
print(f"v * v⁻¹ = {v * v_inv}")  # should yield 1
```

### 8. Random Multivectors

TanGA provides helpers for generating random multivectors:

```python
from pytanga import random_mask, random_mv

# Random blade mask
mask = random_mask(alg, num_blades=3)
print(f"Random blade mask: {mask}")

# Random multivector
rmv = random_mv(alg, mask, value_range=(-1.0, 1.0))
print(f"Random multivector: {rmv}")
print(f"Grades: {rmv.grades}")
```
