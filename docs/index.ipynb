# TanGA Tutorial

Tutorials for the [TanGA](https://github.com/dodeka12/tanga) geometric algebra
library.

This site walks through the **pytanga** Python package in two parts:

- **Visualization** — teaches the `pytanga.viz` viewer from scratch, no
  geometric-algebra background required.
- **Geometric Algebra** — covers the `Algebra` and `MV` types, the basis
  classes, the geometry submodule, and the numerical tooling.


```python
from pytanga.geometry import Point, Sphere
from pytanga.viz import TextureLabelStyle, Visualizer

viz = Visualizer(title="TanGA — Geometric Algebra & Visualization")

# Fully opaque sphere — the texture label is clearly visible.
viz.add(
    Sphere(Point(-1.1, 0, 0), 1.6),
    color="#4488ff",
    opacity=1.0,
    tex_label="$x^2 + y^2 + z^2 = r_1^2$",
    #tex_label_style=TextureLabelStyle(repeat_u=2),
)

# Semi-transparent sphere — lets you see the intersection.
viz.add(
    Sphere(Point(1.1, 0, 0), 1.6),
    color="#ff8844",
    opacity=0.55,
    tex_label=r"$$x^2 + y^2 + z^2 = r_2^2$$",
    tex_label_style=TextureLabelStyle(repeat_u=2),
)

viz.display_snapshot()
```
